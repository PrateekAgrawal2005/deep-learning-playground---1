```python
x = []
y = []

for seq in sequence:
  for i in range(1,len(seq)):
    input_seq = seq[:i]
    output_seq = seq[i]
    x.append(input_seq)
    y.append(output_seq)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, SimpleRNN
```

```python
embedding_dim = 50
```

```python
rnn_units = 128
```

```python
rnn_model = Sequential()
rnn_model.add(Embedding(input_dim = vocab_size, output_dim = embedding_dim))
```

```python
rnn_model.add(SimpleRNN(units = rnn_units))
rnn_model.add(Dense(units = vocab_size, activation = 'softmax'))
```

```python
rnn_model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
rnn_model.summary()
```

```python
lstm_model = Sequential()
```

```python
lstm_model.add(
    Embedding(input_dim = vocab_size, output_dim = embedding_dim)
)
lstm_model.add(LSTM(units = rnn_units))
lstm_model.add(Dense(units = vocab_size, activation = 'softmax'))
```

```python
lstm_model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
def predict_next_word(model, tokenizer, text, max_sequence_len):
    # Convert the input text to sequences of tokens
    token_list = tokenizer.texts_to_sequences([text])[0]

    # Pad the sequence to the appropriate length for the model
    # The model's input length was max_len - 1
    token_list = pad_sequences([token_list], maxlen=max_sequence_len - 1, padding='pre')

    # Predict the probabilities for the next word
    predicted_probs = model.predict(token_list, verbose=0)

    # Get the index of the word with the highest probability
    predicted_word_index = np.argmax(predicted_probs)

    # Map the predicted word index back to a word
    for word, index in tokenizer.word_index.items():
        if index == predicted_word_index:
            return word
    return None

# Example usage:
# Ensure 'lstm_model', 'tokinizer', and 'max_len' are defined and available
# from previous cell executions.
input_text = "the world as we have"
predicted_word = predict_next_word(lstm_model, tokinizer, input_text, max_len)

print(f"Input text: {input_text}")
print(f"Predicted next word: {predicted_word}")

input_text_2 = "love is"
predicted_word_2 = predict_next_word(lstm_model, tokinizer, input_text_2, max_len)
print(f"\nInput text: {input_text_2}")
print(f"Predicted next word: {predicted_word_2}")
```
